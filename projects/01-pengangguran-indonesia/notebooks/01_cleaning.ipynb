### 2. Hipotesis (Pendidikan)

- Serapan industri belum tentu sepadan atau sesuai dengan keahlian yang diajarkan.
- Sebaliknya, lulusan SD/SMP cenderung langsung bekerja apa saja (sektor informal) karena tidak sanggup menunggu pekerjaan yang pas, sehingga TPT-nya kemungkinan lebih rendah.

Hipotesis ini akan diverifikasi dengan data Sakernas di Tahap 2, bukan dianggap sebagai fakta.

### 3. TPT Turun, TPAK Juga Turun
Jika TPT dan TPAK turun bersamaan, kemungkinan terjadi *discouraged worker effect*: sebagian penganggur berhenti mencari kerja dan pindah ke kelompok **bukan angkatan kerja**. Mereka keluar dari penyebut TPT, sehingga angka pengangguran tampak membaik padahal jumlah orang yang bekerja tidak bertambah.

**Ilustrasi (100 penduduk usia kerja):**

| Kondisi | Bekerja | Menganggur | Bukan AK | TPT | TPAK |
|---|---|---|---|---|---|
| Awal | 54 | 6 | 40 | 10% | 60% |
| 3 penganggur berhenti mencari | 54 | 3 | 43 | ±5,3% | 57% |

TPT turun dari 10% ke ±5,3% dan TPAK turun dari 60% ke 57%, tetapi jumlah yang bekerja tetap 54. Jadi tidak ada perbaikan nyata.

**Pengingat:** saat TPT turun, selalu cek TPAK dan jumlah absolut orang yang bekerja sebelum menyimpulkan kondisinya membaik.

## Tahap 3 - Data Cleaning (provinsi - sampel 2026 Feb)
Catatan: file ini adalah sampel dari payload SSR BPS. Struktur sudah rapi.
- Cek duplikasi (kode provinsi)
- Pisahkan baris total 'INDONESIA' (tidak dipakai untuk ranking provinsi)
- Pastikan tipe numerik (tpt float)
- Simpan ke data/clean/tpt_provinsi_clean.csv

```python
import pandas as pd
df = pd.read_csv('../data/raw/tpt_provinsi_bps_embed.csv')
print('shape:', df.shape, 'unik kode:', df['provinsi_kode'].nunique())
df_clean = df[df['provinsi'] != 'INDONESIA'].copy()
df_clean['tpt'] = pd.to_numeric(df_clean['tpt'], errors='coerce')
df_clean.to_csv('../data/clean/tpt_provinsi_clean.csv', index=False)
df_clean.head(5)
```
