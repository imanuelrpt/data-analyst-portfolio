Fase 1 Modul 4: groupby, merge, pivot (terakhir Fase 1)

```python
# groupby

import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv")

print(df.shape)  # run dari C:\Voiceover

# rata-rata bill per hari
print(df.groupby("day")["total_bill"].mean())
# agregasi banyak sekaligus, reset_index biar jadi tabel rapi
rekap = df.groupby("day").agg(total=("total_bill","sum"), rata=("total_bill","mean"), n=("tip","count")).reset_index()
print(rekap)
```

```python
# merge / join (tempel 2 tabel)

# tabel tambahan: kategori hari (data kecil buatan)
hari = pd.DataFrame({"day": ["Thur","Fri","Sat","Sun"], "jenis": ["Weekday","Weekday","Weekend","Weekend"]})
# left join: semua transaksi tetap ada, ditempel jenis harinya
df2 = df.merge(hari, on="day", how="left")
print(df2[["day","jenis","total_bill"]].head(3))
# cek kunci duplikat sebelum merge: hari["day"].duplicated().sum() harus 0
```

```python
# pivot_table (rekap silang)

# rata-rata tip per time x smoker, mirip PivotTable Excel
pivot = df.pivot_table(index="time", columns="smoker", values="tip", aggfunc="mean")
print(pivot)
# kalau ada lubang (NaN), isi 0: .fillna(0)
```
