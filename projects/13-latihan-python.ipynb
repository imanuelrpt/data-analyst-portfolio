```python
from pathlib import Path
import pandas as pd
p = Path("data/clean_tips.csv") if Path("data/clean_tips.csv").exists() else Path("../data/clean_tips.csv")
df = pd.read_csv(p)
df = df[(df["tip"] >= 0) & (df["total_bill"] < 100)]  # 241
# Ronde 1 UNI (1 kolom): df["col"].describe() + value_counts() + 1 hist
# Ronde 2 BI (2 kolom): angka-vs-angka = corr+scatter | kategori-vs-angka = groupby+bar/box
# Ronde 3 MULTI (3+): tambah hue/segment, misal scatter bill-tip + hue Weekend + pisah Lunch/Dinner
```

```python
# UNI angka: sebaran + outlier
print(df["harga_per_orang"].describe(), df["harga_per_orang"].skew().round(2))
# UNI kategori: proporsi (waspada n kecil Fri=19)
print(df["day"].value_counts(), df["day"].value_counts(normalize=True).round(2))
# BI angka-angka: korelasi + scatter (r=0.67 bill-tip kemarin)
print(df[["total_bill","tip"]].corr().round(2))
# BI kategori-angka: Weekend vs Weekday (mean+median+count, jangan mean saja)
print(df.groupby("is_weekend")["total_bayar"].agg(["mean","median","count"]).round(2))
# MULTI: segmen Weekend x bill_level, metrik total_bayar sum (dari Modul 3 Fase 2)
print(df.groupby(["is_weekend","bill_level"])["total_bayar"].sum().round(2))
```
