```python
from pathlib import Path
import pandas as pd, sqlite3
csv = Path("data/clean_tips.csv") if Path("data/clean_tips.csv").exists() else Path("../data/clean_tips.csv")
con = sqlite3.connect(csv.parent / "tips.db")

# SQL -> pandas: agregat di DB (cepat), presentasi di Python
df = pd.read_sql("""SELECT day, COUNT(*) AS n, AVG(total_bayar) AS avg_bill
FROM tips GROUP BY day""", con)
df["share"] = df["n"] / df["n"].sum() * 100  # kolom turunan di Python, bukan SQL
print(df)

# pandas -> SQL: simpan HASIL analisis balik ke DB buat dipakai query lain
df.to_sql("rekap_day", con, if_exists="replace", index=False)
print(pd.read_sql("SELECT * FROM rekap_day ORDER BY n DESC", con))
# Aturan: angka mentah & filter besar di SQL, formatting/plot di Python.
```
