Fase 4 Modul 1: Matplotlib/Seaborn Dasar

```python
# Setup + aturan 1 cell 1 grafik

from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt  # fondasi, semua ikut ini
import seaborn as sns  # cantik + statistik, di atas matplotlib

p = Path("data/clean_tips.csv") if Path("data/clean_tips.csv").exists() else Path("../data/clean_tips.csv")
df = pd.read_csv(p)
df = df[(df["tip"] >= 0) & (df["total_bill"] < 100)]  # 241 bersih, pola lama
print(df.shape)

sns.set_theme(style="whitegrid")  # 1x saja, biar rapi
```

```python
# 1. Histogram: bentuk sebaran (cocok skew 1.12 kemarin)
plt.figure()
plt.hist(df["total_bill"], bins=20)  # bins = jumlah ember
plt.xlabel("total_bill"); plt.ylabel("jumlah transaksi"); plt.title("Sebaran Tagihan")
plt.show()  # di VS Code muncul di output, di .py simpan pakai plt.savefig()

# 2. Boxplot: outlier sekilas per grup
plt.figure()
sns.boxplot(data=df, x="day", y="total_bill")  # x kategori, y angka
plt.title("Tagihan per Hari (titik atas = outlier)")
plt.show()

# 3. Bar mean + Scatter hubungan
plt.figure()
sns.barplot(data=df, x="day", y="tip_pct", estimator="median")  # median, bukan mean (miring)
plt.title("Median Royalitas per Hari"); plt.show()

plt.figure()
sns.scatterplot(data=df, x="total_bill", y="tip")  # korelasi 0.67 kemarin, harusnya naik
plt.title("Bill vs Tip"); plt.show()
```
