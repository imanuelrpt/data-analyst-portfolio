Fase 4 Modul 2: Pilih Chart + Storytelling

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
p = Path("data/clean_tips.csv") if Path("data/clean_tips.csv").exists() else Path("../data/clean_tips.csv")
df = pd.read_csv(p)
df = df[(df["tip"] >= 0) & (df["total_bill"] < 100)]  # 241

# contoh storytelling: bar dengan judul ber-insight + anotasi angka
med = df.groupby("day")["tip_pct"].median().reindex(["Thur","Fri","Sat","Sun"])
fig, ax = plt.subplots(figsize=(7, 4))
sns.barplot(x=med.index, y=med.values, ax=ax)
ax.bar_label(ax.containers[0], fmt="%.1f")
ax.set(title="Fri paling royal 17.0% tapi cuma 19 transaksi -> jangan promo ke sini", ylabel="median tip_pct")
plt.savefig("story_bar.png", dpi=150, bbox_inches="tight"); plt.close()
print("simpan story_bar.png")
```
