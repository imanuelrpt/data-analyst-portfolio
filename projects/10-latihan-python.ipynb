```python
from pathlib import Path
import pandas as pd
p = Path("data/clean_tips.csv") if Path("data/clean_tips.csv").exists() else Path("../data/clean_tips.csv")
df = pd.read_csv(p)
df = df[(df["tip"] >= 0) & (df["total_bill"] < 100)]  # buang 2 error, pola lama
print(df.shape)  # 241
```

```python
from scipy.stats import ttest_ind  # banding mean 2 grup independen
weekend = df[df["is_weekend"]==True]["total_bayar"]
weekday = df[df["is_weekend"]==False]["total_bayar"]
stat, p = ttest_ind(weekend, weekday, equal_var=False)  # Welch, aman varians beda
print(f"mean weekend {weekend.mean():.2f} vs weekday {weekday.mean():.2f}, p={p:.4f}")
# baca: p<0.05 = beda signifikan, bukan kebetulan sampel. Tapi cek selisih rupiahnya bermakna bisnis tidak.
```

```python
# A = Weekday, B = Weekend. Metrik: tip_pct. H0: mean sama.
from scipy.stats import ttest_ind
a = df[df["is_weekend"]==False]["tip_pct"]
b = df[df["is_weekend"]==True]["tip_pct"]
stat, p = ttest_ind(a, b, equal_var=False)
print(f"A {a.mean():.2f}% vs B {b.mean():.2f}%, p={p:.4f}")
# keputusan: p>=0.05 = belum cukup bukti B lebih royal. Jangan launch promo berbasis beda 0.1% kemarin.
```
