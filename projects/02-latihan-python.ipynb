FASE 1 MODUL 2: NumPy — Excel 1 Kolom yang Super Cepat

```python
# Array Dasar

import numpy as np  # standar alias, jangan diganti

# buat array dari list
omzet = np.array([20000, 15000, 45000, 30000])
print(omzet)  # array([20000, 15000, 45000, 30000])
print(omzet.dtype)  # tipe data seragam, misal int64
print(omzet.shape)  # (4,) = 1 dimensi, 4 elemen

# cara cepat bikin array
print(np.zeros(3))  # [0. 0. 0.] buat inisialisasi
print(np.arange(0, 10, 2))  # [0 2 4 6 8] mirip range()
```

```python
# Operasi tanpa loop (vectorized)

omzet = np.array([20000, 15000, 45000, 30000])

print(omzet * 2)  # semua elemen dikali 2 sekaligus
print(omzet - 5000)  # diskon 5000 ke semua

# agregasi = jawaban 1 angka
print(omzet.sum())  # total
print(omzet.mean())  # rata-rata
print(omzet.max(), omzet.min())  # max, min
print(omzet.std())  # standar deviasi, sebaran data
```

```python
#  boolean masking (filter)

omzet = np.array([20000, 15000, 45000, 30000])

mask = omzet > 25000  # array boolean [False False True True]
print(mask)

print(omzet[mask])  # ambil yang True saja -> [45000 30000]
print(omzet[omzet < 20000])  # langsung 1 baris, tanpa loop/if
```

```python
# Array 2D + axis

# baris = kota, kolom = minggu
# Jakarta: [10, 12, 8], Bandung: [5, 7, 6]
data = np.array([
    [10, 12, 8],
    [5, 7, 6]
])
print(data.shape)  # (2, 3) = 2 baris, 3 kolom

print(data.sum(axis=0))  # axis=0 = ke bawah, total per minggu -> [15 19 14]
print(data.sum(axis=1))  # axis=1 = ke samping, total per kota -> [30 18]
# trik ingat: axis yang disebut itu yang hilang. sum axis=1, kolom hilang, sisa per baris.
```
