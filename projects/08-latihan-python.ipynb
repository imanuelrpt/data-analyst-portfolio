Fase 3 Modul 1: Mean, Median, Modus, Sebaran, Distribusi

```python
from pathlib import Path
import pandas as pd

p = Path("data/clean_tips.csv") if Path("data/clean_tips.csv").exists() else Path("../data/clean_tips.csv")
print(f"pakai: {p}")
df = pd.read_csv(p)
print(df.shape)  # harusnya (243, 17) versi dirty2

# tiga tengah untuk tip_pct (royalitas %)
print(df["tip_pct"].mean())  # ketarik outlier 71% dan -48%
print(df["tip_pct"].median())  # titik tengah, aman
print(df["day"].mode()[0])  # kategori pakai modus: Sat
# bukti miring: mean > median = miring kanan (banyak kecil, sedikit besar)
```

```python
# Sebaran: varians & standar deviasi

# dua menu bisa mean sama tapi sebar beda: satu konsisten, satu kadang kecil kadang jumbo
print(df["tip_pct"].std())  # simpangan baku: rata-rata jarak ke mean
print(df["tip_pct"].quantile([0.25, 0.5, 0.75]))  # Q1, median, Q3
print(df["tip_pct"].describe())  # baca min-max-std sekilas
# aturan praktis: std besar vs mean = data menyebar, jangan lapor mean saja
```

```python
# Bentuk distribusi (tanpa grafik mewah)

print(df["tip_pct"].skew())  # >0 miring kanan, <0 miring kiri, ~0 simetris
# cek ekor: berapa % di atas mean + 2*std? (calon outlier)
batas = df["tip_pct"].mean() + 2*df["tip_pct"].std()
print(f"batas: {batas:.1f}, jumlah di atas: {(df['tip_pct'] > batas).sum()}")
```
