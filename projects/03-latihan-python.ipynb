```python
 # Baca + intip (wajib tiap mulai kerja)

import pandas as pd  # alias baku, jangan diganti

# baca langsung dari URL, tanpa download manual
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)  # hasil = DataFrame

print(df.shape)  # (baris, kolom) -> (244, 7)
print(df.head(3))  # 3 baris pertama, cek cepat isi
print(df.info())  # kolom apa, tipe apa, ada null tidak
print(df.describe())  # statistik cepat kolom angka saja
```

```python
# Ambil kolom / baris

print(df["total_bill"].head())  # 1 kolom = Series
print(df[["total_bill", "tip"]].head())  # 2 kolom = DataFrame mini, pakai list

# 1 baris pakai iloc (posisi), loc (filter)
print(df.iloc[0])  # baris ke-0
```

```python
#  Filter (ini 80% kerja analyst)

# pola: df[ kondisi ]
mahal = df[df["total_bill"] > 40]  # WHERE total_bill > 40
print(mahal.shape)  # berapa baris yang lolos?

# 2 syarat: pakai & (dan), | (atau), tiap syarat dalam kurung
weekend_dinner = df[(df["day"].isin(["Sat", "Sun"])) & (df["time"] == "Dinner")]
print(weekend_dinner.head())

# cara aman ubah hasil filter (hindari SettingWithCopyWarning):
# pakai .copy() kalau mau diolah lanjut
tinggi = df[df["tip"] > 5].copy()
tinggi["tip_double"] = tinggi["tip"] * 2  # tambah kolom baru, aman karena .copy()
```

```python
# Sorting

# urutkan tagihan terbesar ke terkecil
top5 = df.sort_values("total_bill", ascending=False).head(5)
print(top5[["total_bill", "tip", "day"]])

# sort 2 kolom: hari naik, tagihan turun
print(df.sort_values(["day", "total_bill"], ascending=[True, False]).head())
```
