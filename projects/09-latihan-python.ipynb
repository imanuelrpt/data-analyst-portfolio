Fase 3 Modul 2: Korelasi, Sampling, Confidence Interval

```python
# Korelasi (hubungan, bukan sebab)

from pathlib import Path
import pandas as pd

p = Path("data/clean_tips.csv") if Path("data/clean_tips.csv").exists() else Path("../data/clean_tips.csv")
df = pd.read_csv(p)
# buang 2 error input biar tidak menipu (tip<0, bill raksasa) - pola Modul 1
df = df[(df["tip"] >= 0) & (df["total_bill"] < 100)]

# matriks korelasi kolom angka saja, -1 sampai 1
print(df[["total_bill","tip","size","tip_pct"]].corr().round(2))
# baca: total_bill vs tip ~0.6-0.7 = gerak bareng sedang-kuat
# tip_pct vs total_bill negatif kecil = tagihan besar, persen tip turun (cocok temuan Low 18% vs High 13%)
# INGAT: korelasi != sebab. tip besar tidak bikin bill besar, keduanya naik bareng karena makan banyak.
```

```python
# Sampling

# populasi 241 baris, sampel 30 acak dengan seed biar bisa diulang
sampel = df.sample(n=30, random_state=42)  # random_state = hasil sama tiap run
print(sampel["total_bayar"].mean(), df["total_bayar"].mean())  # dekat tapi tidak sama
# salah umum: sampel tanpa seed -> tiap run beda, tidak bisa diaudit
```

```python
# Confidence Interval 95%

import numpy as np
# CI mean: mean ± 1.96 * std/sqrt(n) (n besar, aproksimasi normal)
x = df["total_bayar"]
mean, se = x.mean(), x.std()/np.sqrt(len(x))
print(f"mean {mean:.2f}, CI95% [{mean-1.96*se:.2f}, {mean+1.96*se:.2f}]")
# baca: "rata-rata bayar populasi kemungkinan 95% ada di rentang ini"
# n kecil -> rentang lebar (tidak yakin), n besar -> sempit
```
