```python
# 1 transaksi = 1 dict (mirip 1 baris Excel)
transaksi = {"menu": "Ayam Goreng", "harga": 15000, "qty": 2}
print(transaksi["menu"])  # ambil value pakai key
print(transaksi.get("diskon", 0))  # aman kalau key tidak ada, default 0

# 1 hari = list of dict (mirip 1 sheet)
semua = [
    {"menu": "Ayam Goreng", "harga": 15000, "qty": 2},
    {"menu": "Tempe", "harga": 5000, "qty": 5},
    {"menu": "Ayam Goreng", "harga": 15000, "qty": 1},
]

# loop + if = filter & total (nanti jadi df[df.harga > ...])
total = 0  # akumulator, pola wajib
for t in semua:  # t = tiap dict
    if t["menu"] == "Ayam Goreng":  # filter
        total += t["harga"] * t["qty"]  # agregasi
print(total)  # harusnya 45000
```

```python
# function = biar bisa dipakai ulang, jangan copy-paste
def total_pendapatan(data, menu_cari):
    """Hitung total pendapatan untuk 1 menu."""
    total = 0
    for t in data:
        if t["menu"] == menu_cari:
            total += t["harga"] * t["qty"]
    return total

print(total_pendapatan(semua, "Tempe"))  # 25000
```

```python
# list comprehension = loop versi 1 baris, wajib di pandas nanti
# pola: [hasil for item in list if syarat]
harga_mahal = [t["menu"] for t in semua if t["harga"] > 10000]
print(harga_mahal)  # ['Ayam Goreng', 'Ayam Goreng']

qty_besar = [t["qty"] * 2 for t in semua]  # transformasi
print(qty_besar)
```

```python
penjualan = [
    {"kota": "Jakarta", "produk": "Kopi", "qty": 10, "harga": 20000},
    {"kota": "Bandung", "produk": "Teh", "qty": 5, "harga": 15000},
    {"kota": "Jakarta", "produk": "Teh", "qty": 3, "harga": 15000},
    {"kota": "Surabaya", "produk": "Kopi", "qty": 7, "harga": 20000},
]

# 1. (Mudah) Total semua qty
def total_qty(data):
    return sum(item["qty"] for item in data)

print(total_qty(penjualan))


# 2. (Sedang) Kota yang beli Kopi, pakai list comprehension
kota_kopi = [item["kota"] for item in penjualan if item["produk"] == "Kopi"]
print(kota_kopi)


# 3. (Sedang+) Omzet per kota tertentu
def omzet_per_kota(data, kota):
    return sum(item["qty"] * item["harga"] for item in data if item["kota"] == kota)

print(omzet_per_kota(penjualan, "Jakarta")) 


# 4. (Sulit) Rekap omzet semua kota pakai loop + dict
rekap = {}
for item in penjualan:
    kota = item["kota"]
    omzet = item["qty"] * item["harga"]
    if kota in rekap:
        rekap[kota] += omzet
    else:
        rekap[kota] = omzet

print(rekap)
```
